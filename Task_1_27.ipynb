```python
# Task 8: Find the area of a triangle using Heron's formula

# The three sides of the triangle
a = 5
b = 6
c = 7

# Uncomment the lines below to take the sides from the user
# a = float(input('Enter first side: '))
# b = float(input('Enter second side: '))
# c = float(input('Enter third side: '))

# Calculate the semi-perimeter (half of the perimeter)
s = (a + b + c) / 2

# Heron's formula: area = square root of s(s-a)(s-b)(s-c)
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5

# Print the area with 2 decimal places
print('The area of the triangle is %0.2f' % area)
```

```python
# Task 9: Solve the quadratic equation ax**2 + bx + c = 0

# cmath is used so the program also works when the roots are complex
import cmath

# Coefficients of the equation: x**2 + 5x + 6 = 0
a = 1
b = 5
c = 6

# Calculate the discriminant: d = b^2 - 4ac
d = (b ** 2) - (4 * a * c)

# Find the two solutions using the quadratic formula
sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

# Display both solutions
print('The solution are {0} and {1}'.format(sol1, sol2))
```

```python
# Task 15: Exercises on for loops

# Exercise 1: loop over a list
print("Exercise 1: loop over a list")
fruits = ["apple", "banana", "mango"]
for fruit in fruits:           # "fruit" takes each value of the list one by one
    print(fruit)

# Exercise 2: loop with range(); range(1, 6) gives 1,2,3,4,5 (6 is excluded)
print("\nExercise 2: range(1, 6)")
for n in range(1, 6):
    print(n)

# Exercise 3: loop over a string (one character at a time)
print("\nExercise 3: loop over a string")
for ch in "AI":
    print(ch)

# Exercise 4: sum of numbers from 1 to 10
print("\nExercise 4: sum of 1 to 10")
total = 0
for n in range(1, 11):
    total = total + n          # add the current number to total
print("Sum =", total)

# Exercise 5: multiplication table of 5
print("\nExercise 5: table of 5")
for n in range(1, 6):
    print("5 x", n, "=", 5 * n)

# Exercise 6: nested for loop (star pattern)
print("\nExercise 6: star pattern")
for row in range(1, 4):
    for col in range(row):
        print("*", end=" ")     # end=" " keeps the output on the same line
    print()                     # move to the next line after each row
```

```python
# Task 16: Exercises on while loops

# Exercise 1: print numbers 1 to 5
print("Exercise 1: count 1 to 5")
count = 1
while count <= 5:              # loop runs as long as the condition is True
    print(count)
    count = count + 1          # increase count, otherwise the loop never ends

# Exercise 2: countdown from 5 to 1
print("\nExercise 2: countdown")
n = 5
while n > 0:
    print(n)
    n = n - 1

# Exercise 3: sum of the first 10 numbers
print("\nExercise 3: sum of 1 to 10")
i = 1
total = 0
while i <= 10:
    total = total + i
    i = i + 1
print("Sum =", total)

# Exercise 4: while-else (else runs when the loop finishes normally)
print("\nExercise 4: while-else")
x = 1
while x <= 3:
    print("x =", x)
    x = x + 1
else:
    print("Loop finished")
```

```python
# Task 17: Exercises on the break statement

# Exercise 1: stop a for loop when the value is 3
print("Exercise 1: break in a for loop")
for n in range(1, 10):
    if n == 3:
        break                  # exit the loop immediately
    print(n)

# Exercise 2: stop a while loop when the number is found
print("\nExercise 2: break in a while loop")
i = 0
while True:                    # infinite loop, stopped only by break
    i = i + 1
    if i == 4:
        break
    print("i =", i)

# Exercise 3: search an item in a list and stop once it is found
print("\nExercise 3: search with break")
names = ["Ali", "Sara", "Ahmed", "Zain"]
for name in names:
    if name == "Ahmed":
        print("Found Ahmed")
        break
    print("Checked", name)

# Exercise 4: break only exits the INNER loop in nested loops
print("\nExercise 4: break in nested loops")
for a in range(1, 4):
    for b in range(1, 4):
        if b == 2:
            break              # exits only the inner loop
        print(a, b)
```

```python
# Task 18: Exercises on the continue statement

# Exercise 1: skip the number 3 (continue jumps to the next iteration)
print("Exercise 1: skip 3")
for n in range(1, 6):
    if n == 3:
        continue               # skip the rest of this iteration
    print(n)

# Exercise 2: print only odd numbers from 1 to 10
print("\nExercise 2: odd numbers")
for n in range(1, 11):
    if n % 2 == 0:             # even number: skip it
        continue
    print(n)

# Exercise 3: continue in a while loop
print("\nExercise 3: continue in while")
i = 0
while i < 5:
    i = i + 1                  # update first, so the loop does not get stuck
    if i == 2:
        continue
    print("i =", i)

# Exercise 4: skip vowels in a word
print("\nExercise 4: skip vowels")
for ch in "learning":
    if ch in "aeiou":
        continue
    print(ch, end="")
print()
```

```python
# Task 19: Looping techniques in Python

# Technique 1: enumerate() gives index and value together
print("Technique 1: enumerate")
colors = ["red", "green", "blue"]
for index, color in enumerate(colors):
    print(index, color)

# Technique 2: zip() loops over two lists at the same time
print("\nTechnique 2: zip")
names = ["Ali", "Sara"]
marks = [85, 92]
for name, mark in zip(names, marks):
    print(name, "scored", mark)

# Technique 3: items() loops over a dictionary (key and value)
print("\nTechnique 3: dictionary items()")
student = {"name": "Ali", "age": 20}
for key, value in student.items():
    print(key, "->", value)

# Technique 4: sorted() loops in sorted order (original list stays unchanged)
print("\nTechnique 4: sorted")
for n in sorted([4, 1, 3]):
    print(n)

# Technique 5: reversed() loops in reverse order
print("\nTechnique 5: reversed")
for n in reversed([1, 2, 3]):
    print(n)

# Technique 6: range() with a step value (start, stop, step)
print("\nTechnique 6: range with step")
for n in range(0, 10, 3):
    print(n)
```

```python
# Task 20: Exercises on user defined functions

# Exercise 1: simple function without parameters
def greet():
    print("Hello from a function!")

greet()                        # calling the function

# Exercise 2: function with parameters and a return value
def add(a, b):
    return a + b               # return sends the result back to the caller

print("add(3, 4) =", add(3, 4))

# Exercise 3: default argument value
def power(base, exp=2):        # exp is 2 if no value is given
    return base ** exp

print("power(5) =", power(5))
print("power(2, 3) =", power(2, 3))



# Exercise 4: **kwargs accepts any number of keyword arguments
def show_info(**info):
    for key, value in info.items():
        print(key, ":", value)

show_info(name="Ali", age=20)

# Exercise 5: recursive function (a function that calls itself)
def factorial(n):
    if n == 0 or n == 1:       # base case stops the recursion
        return 1
    return n * factorial(n - 1)

print("factorial(5) =", factorial(5))
```

```python
# Task 21: Exercises on the List data type

# Create a list
nums = [10, 20, 30, 40]
print("List:", nums)

# Indexing: first item is index 0, last item is index -1
print("First:", nums[0], "| Last:", nums[-1])

# Slicing: items from index 1 up to (not including) 3
print("Slice nums[1:3]:", nums[1:3])

# append(): add an item at the end
nums.append(50)
print("After append:", nums)

# insert(): add an item at a specific position (index, value)
nums.insert(1, 15)
print("After insert:", nums)

# remove(): delete the first matching value
nums.remove(30)
print("After remove(30):", nums)

# pop(): remove and return the item at an index (last item by default)
last = nums.pop()
print("Popped:", last, "| List:", nums)

# extend(): add all items of another list
nums.extend([1, 2])
print("After extend:", nums)

# sort() and reverse()
nums.sort()
print("Sorted:", nums)
nums.reverse()
print("Reversed:", nums)

# Restore the built-in sum function in case it was shadowed by previous tasks
try:
    import builtins
    sum = builtins.sum
except Exception:
    pass

# len(), min(), max(), sum()
print("Length:", len(nums), "| Min:", min(nums), "| Max:", max(nums), "| Sum:", sum(nums))

# List comprehension: squares of 1 to 5
squares = [n ** 2 for n in range(1, 6)]
print("Squares:", squares)
```

```python
# Task 22: Exercises on the Tuple data type

# Create a tuple (uses round brackets and cannot be changed after creation)
t = (10, 20, 30, 20)
print("Tuple:", t)

# Indexing and slicing work like lists
print("First:", t[0], "| Last:", t[-1])
print("Slice t[1:3]:", t[1:3])

# count(): how many times a value appears
print("Count of 20:", t.count(20))

# index(): position of the first matching value
print("Index of 30:", t.index(30))

# len()
print("Length:", len(t))

# A tuple with ONE item needs a trailing comma
single = (5,)
print("Single item tuple:", single, type(single))

# Unpacking: assign tuple values to separate variables
a, b, c, d = t
print("Unpacked:", a, b, c, d)

# Concatenation and repetition
print("Joined:", (1, 2) + (3, 4))
print("Repeated:", (1, 2) * 2)

# Tuples are immutable: changing a value gives an error
try:
    t[0] = 99
except TypeError as e:
    print("Error:", e)

# Convert between list and tuple to "modify" a tuple
temp = list(t)
temp[0] = 99
t = tuple(temp)
print("Modified tuple:", t)
```

```python
# Task 23: Exercises on the String data type

s = "  Machine Learning  "
print("Original:", repr(s))

# strip(): remove spaces from both ends
s = s.strip()
print("strip():", s)

# upper(), lower(), title()
print("upper():", s.upper())
print("lower():", s.lower())
print("title():", s.title())

# len(): number of characters (spaces count too)
print("Length:", len(s))

# Indexing and slicing
print("First char:", s[0], "| Last char:", s[-1])
print("Slice s[0:7]:", s[0:7])

# Reverse a string using slicing with step -1
print("Reversed:", s[::-1])

# replace(): change a part of the string
print("replace():", s.replace("Machine", "Deep"))

# find(): position of a word (-1 if not found)
print("find('Learning'):", s.find("Learning"))

# split(): break a string into a list of words
words = s.split()
print("split():", words)

# join(): combine a list into a string
print("join():", "-".join(words))

# startswith() and endswith()
print("startswith('Mac'):", s.startswith("Mac"))
print("endswith('ing'):", s.endswith("ing"))

# String formatting: format() and f-strings
name = "Ali"
age = 20
print("{} is {} years old".format(name, age))
print(f"{name} is {age} years old")

# Check if a word is inside a string
print("'Learning' in s:", "Learning" in s)
```

```python
# Task 24: Exercises on the Set data type

# Create a set (unordered, no duplicate values)
s = {1, 2, 3, 3, 2}
print("Set (duplicates removed):", s)

# add(): add one item
s.add(4)
print("After add(4):", s)

# update(): add many items
s.update([5, 6])
print("After update:", s)

# remove(): delete an item (error if missing), discard(): delete without error
s.remove(6)
s.discard(100)                 # 100 is not in the set, but no error occurs
print("After remove/discard:", s)

# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union (a | b):", a | b)
print("Intersection (a & b):", a & b)
print("Difference (a - b):", a - b)
print("Symmetric difference (a ^ b):", a ^ b)

# Subset check
print("{1, 2} is subset of a:", {1, 2}.issubset(a))

# Use a set to remove duplicates from a list
data = [1, 2, 2, 3, 3, 3]
print("Unique values:", list(set(data)))

# Membership test
print("2 in a:", 2 in a)

# len()
print("Length of a:", len(a))
```

```python
# Task 25: Exercises on the Dictionary data type

# Create a dictionary (key: value pairs)
student = {"name": "Ali", "age": 20, "course": "AI"}
print("Dictionary:", student)

# Access a value by its key
print("Name:", student["name"])

# get(): safe access (returns None or a default if the key is missing)
print("Marks:", student.get("marks", "Not available"))

# Add a new key and update an existing key
student["marks"] = 90
student["age"] = 21
print("After add/update:", student)

# pop(): remove a key and return its value
removed = student.pop("course")
print("Removed:", removed)

# del: delete a key
del student["marks"]
print("After del:", student)

# keys(), values(), items()
print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
print("Items:", list(student.items()))

# Loop through a dictionary
for key, value in student.items():
    print(key, "->", value)

# Check if a key exists
print("'name' in student:", "name" in student)

# Nested dictionary
school = {"s1": {"name": "Ali"}, "s2": {"name": "Sara"}}
print("Nested access:", school["s2"]["name"])

# Dictionary comprehension: number -> square
squares = {n: n ** 2 for n in range(1, 6)}
print("Squares:", squares)
```

```python
# Task 26: Exercises on Exception Handling

# Exercise 1: basic try-except (division by zero)
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: cannot divide by zero")

# Exercise 2: ValueError (wrong value for conversion)
try:
    number = int("abc")
except ValueError:
    print("Error: 'abc' is not a valid integer")

# Exercise 3: multiple except blocks and the exception message
try:
    data = [1, 2, 3]
    print(data[5])
except IndexError as e:
    print("IndexError:", e)
except Exception as e:
    print("Other error:", e)

# Exercise 4: try-except-else-finally
try:
    value = int("25")
except ValueError:
    print("Invalid number")
else:
    # Runs only when NO exception occurred
    print("Converted successfully:", value)
finally:
    # Always runs, whether an error happened or not
    print("This line always runs")

# Exercise 5: FileNotFoundError
try:
    with open("missing_file.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("Error: file not found")

# Exercise 6: raise an exception manually
try:
    age = -5
    if age < 0:
        raise ValueError("Age cannot be negative")
except ValueError as e:
    print("Error:", e)
```

```python
# Task 27: Exercises on User Defined Exception Handling

# Exercise 1: create your own exception class by inheriting from Exception
class NegativeAgeError(Exception):
    """Raised when the age is negative."""
    pass

def set_age(age):
    if age < 0:
        raise NegativeAgeError("Age cannot be negative")
    return age

try:
    set_age(-3)
except NegativeAgeError as e:
    print("Caught NegativeAgeError:", e)

# Exercise 2: custom exception with extra data (balance example)
class InsufficientBalanceError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        # Build a clear message and pass it to the parent class
        super().__init__(f"Balance {balance} is less than the requested amount {amount}")

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError(balance, amount)
    return balance - amount

try:
    withdraw(500, 1000)
except InsufficientBalanceError as e:
    print("Caught InsufficientBalanceError:", e)
    print("Short by:", e.amount - e.balance)

# Exercise 3: a valid case runs normally
print("New balance:", withdraw(1000, 300))
```
