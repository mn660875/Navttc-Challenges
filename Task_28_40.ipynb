```python
# Task 28: Create NumPy arrays
import numpy as np

# 1D array from a Python list
a = np.array([1, 2, 3, 4, 5])
print("1D array:", a)

# 2D array (rows and columns)
b = np.array([[1, 2, 3], [4, 5, 6]])
print("2D array:\n", b)

# Arrays of zeros and ones
print("Zeros:", np.zeros(3))
print("Ones:", np.ones(3))
```

```python
# Task 29: NumPy indexing
import numpy as np

a = np.array([10, 20, 30, 40, 50])

# Index starts from 0
print("First element:", a[0])
print("Third element:", a[2])

# Negative index counts from the end
print("Last element:", a[-1])

# 2D array: array[row, column]
b = np.array([[1, 2, 3], [4, 5, 6]])
print("Row 1, Column 2:", b[1, 2])
```

```python
# Task 30: NumPy slicing
import numpy as np

a = np.array([10, 20, 30, 40, 50, 60])

# Slice format: array[start:end]  (end is not included)
print("a[1:4]:", a[1:4])

# From the start up to index 3
print("a[:3]:", a[:3])

# From index 3 to the end
print("a[3:]:", a[3:])

# Every 2nd element: array[start:end:step]
print("a[::2]:", a[::2])
```

```python
# Task 31: NumPy data types
import numpy as np

# Check the data type of an array
a = np.array([1, 2, 3])
print("Integer array dtype:", a.dtype)

b = np.array([1.5, 2.5, 3.5])
print("Float array dtype:", b.dtype)

# Create an array with a chosen data type
c = np.array([1, 2, 3], dtype="float")
print("Chosen dtype:", c.dtype, c)

# Change the data type of an existing array
d = a.astype("float")
print("After astype:", d.dtype, d)
```

```python
# Task 32: NumPy copy and view
import numpy as np

# copy() makes a NEW array, so changing the original does not affect it
a = np.array([1, 2, 3])
c = a.copy()
a[0] = 100
print("Original:", a)
print("Copy:    ", c)

# view() shares the SAME data, so changing the original also changes the view
b = np.array([1, 2, 3])
v = b.view()
b[0] = 100
print("Original:", b)
print("View:    ", v)
```

```python
# Task 33: NumPy array shape
import numpy as np

a = np.array([1, 2, 3, 4])
print("Shape of 1D array:", a.shape)

b = np.array([[1, 2, 3], [4, 5, 6]])
print("Shape of 2D array:", b.shape)    # (rows, columns)

# Other useful properties
print("Dimensions:", b.ndim)
print("Total elements:", b.size)
```

```python
# Task 34: NumPy reshape
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6])

# Change 1D array (6 items) into 2 rows and 3 columns
b = a.reshape(2, 3)
print("Reshaped to (2, 3):\n", b)

# Change it into 3 rows and 2 columns
c = a.reshape(3, 2)
print("Reshaped to (3, 2):\n", c)

# Flatten back to 1D
print("Flattened:", c.reshape(-1))
```

```python
# Task 35: NumPy array iteration
import numpy as np

# Iterate over a 1D array
a = np.array([1, 2, 3])
for x in a:
    print(x)

# Iterate over a 2D array (each loop gives one row)
b = np.array([[1, 2], [3, 4]])
for row in b:
    print("Row:", row)

# Iterate over every single element of a 2D array
for x in np.nditer(b):
    print(x, end=" ")
```

```python
# Task 36: Join (concatenate) NumPy arrays
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Join two 1D arrays
print("Joined:", np.concatenate((a, b)))

# Join two 2D arrays row-wise (axis=0) and column-wise (axis=1)
x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])
print("Axis 0 (below):\n", np.concatenate((x, y), axis=0))
print("Axis 1 (side by side):\n", np.concatenate((x, y), axis=1))
```

```python
# Task 37: Split a NumPy array
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6])

# Split into 3 equal parts (the result is a list of arrays)
parts = np.array_split(a, 3)
print("Parts:", parts)

# Access one part
print("First part:", parts[0])
```

```python
# Task 38: Search in a NumPy array
import numpy as np

a = np.array([10, 20, 30, 20, 40])

# where(): find the index positions of a value
print("Positions of 20:", np.where(a == 20))

# Find all values greater than 25
print("Values > 25:", a[a > 25])

# searchsorted(): where to insert a value in a SORTED array
sorted_a = np.array([1, 3, 5, 7])
print("Insert 4 at index:", np.searchsorted(sorted_a, 4))
```

```python
# Task 39: Sort a NumPy array
import numpy as np

# Sort numbers (small to large)
a = np.array([3, 1, 4, 2])
print("Sorted numbers:", np.sort(a))

# Sort text (alphabetical order)
b = np.array(["mango", "apple", "banana"])
print("Sorted text:", np.sort(b))

# Sort a 2D array (each row is sorted)
c = np.array([[3, 2, 1], [9, 7, 8]])
print("Sorted 2D:\n", np.sort(c))
```

```python
# Task 40: NumPy random numbers
import numpy as np

# Fix the seed so the same numbers come every time you run the cell
np.random.seed(42)

# Random integer from 0 to 99
print("Random integer:", np.random.randint(100))

# Array of 5 random integers from 1 to 10
print("5 random integers:", np.random.randint(1, 11, size=5))

# Random float between 0 and 1
print("Random float:", np.random.rand())

# Pick a random value from a given list
print("Random choice:", np.random.choice([10, 20, 30]))
```
