```python
# Task 43: Pandas Series
import pandas as pd

# Create a Series from a list (index starts from 0)
s = pd.Series([10, 20, 30])
print(s)

# Access one value by its index
print("Value at index 1:", s[1])

# Create a Series with your own labels (index)
marks = pd.Series([85, 90, 70], index=["Ali", "Sara", "Ahmed"])
print(marks)

# Access a value by its label
print("Sara's marks:", marks["Sara"])
```

```python
# Task 44: Pandas DataFrame
import pandas as pd

# Create a DataFrame from a dictionary (keys become column names)
data = {
    "name": ["Ali", "Sara", "Ahmed"],
    "age": [20, 21, 22],
    "marks": [85, 90, 70]
}
df = pd.DataFrame(data)
print(df)

# Select one column
print("\nNames column:")
print(df["name"])

# Select one row by its position using loc
print("\nFirst row:")
print(df.loc[0])
```

```python
# Task 46: Analyze data with Pandas
import pandas as pd

df = pd.read_csv("data.csv")

# shape: (number of rows, number of columns)
print("Shape:", df.shape)

# info(): column names, data types, and non-null counts
df.info()

# describe(): count, mean, min, max, etc. of the numeric columns
print(df.describe())

# Average of one column
print("Average marks:", df["marks"].mean())

# Highest value of one column
print("Highest hours:", df["hours"].max())
```

```python
# Task 47: Clean data with Pandas
import pandas as pd

df = pd.read_csv("data.csv")

# Step 1: count missing values in each column
print("Missing values:")
print(df.isnull().sum())

# Step 2: remove duplicate rows (row 5 is a copy of row 0)
df = df.drop_duplicates()
print("\nRows after removing duplicates:", len(df))

# Step 3: fill the missing marks with the average marks
df["marks"] = df["marks"].fillna(df["marks"].mean())

# Check the cleaned data
print("\nCleaned data:")
print(df)
```

```python
# Task 48: Correlation with Pandas
import pandas as pd

df = pd.read_csv("data.csv")

# Clean the data first (same steps as Task 47)
df = df.drop_duplicates()
df["marks"] = df["marks"].fillna(df["marks"].mean())

# corr() shows how strongly numeric columns are related.
# Value near +1 = increase together, near -1 = one increases while the other decreases,
# near 0 = no relation.
print(df[["age", "marks", "hours"]].corr())

# Correlation between just two columns
print("Correlation (hours, marks):", df["hours"].corr(df["marks"]))
```
